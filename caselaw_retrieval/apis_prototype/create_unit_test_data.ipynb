```python
import json
import os
from facilex_platform_code_case_recommendation import *
```

```python
path = "testing_data/input"
os.makedirs(path, exist_ok=True)
corpus = json.load(open("corpus.json", "r"))

# create input data
for idx, case in enumerate(corpus):
    json.dump(case, open(path + f"/{idx}.json", "w"), indent=2)
```

```python
from tqdm import tqdm

path_expected = "testing_data/expected"
os.makedirs(path_expected, exist_ok=True)
for input_case in tqdm(os.listdir(path)):
    similar_cases = semantic_search(path + "/" + input_case)
    json.dump(similar_cases, open(path_expected + "/" + f"expected_{input_case}", "w"))
```
